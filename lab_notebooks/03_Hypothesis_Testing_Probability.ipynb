## Hypothesis Testing

**What is hypothesis testing?**

Hypothesis testing is a method for making decisions using data. It helps us determine whether a pattern we observe is likely due to random chance, or whether something real is going on.

The core idea: assume a boring baseline is true, then ask how likely our data would be if that baseline really were true. If the data would be very unlikely under the baseline, we have evidence against it.

**Why is it useful?**

Hypothesis testing lets us:
- Evaluate claims about data ("Is this feature actually useful for my model?")
- Compare groups ("Do students who study more actually get better grades?")
- Decide whether a result is statistically significant or just noise

---

### Null and Alternative Hypotheses

Every hypothesis test has two competing claims:

- **Null Hypothesis (H0):** The "nothing interesting is happening" explanation. Any pattern we observe is due to chance.
- **Alternative Hypothesis (H1):** Something real is happening. There is an actual effect or difference.

Tips:
- The null is always the default or status quo.
- The alternative is what you actually suspect might be true.
- Both should refer to a specific, measurable quantity (like a mean or proportion).

Example:
- H0: The average height of students is 5'7" (the national average).
- H1: The average height of students is different from 5'7".

### Q4 — P-Values

The **p-value** is the probability of observing a test statistic as extreme as (or more extreme than) our observed value, assuming the null hypothesis is true.

- A **small** p-value (below the significance level) means our data is very unlikely under H0. We reject H0.
- A **large** p-value means our data is plausible under H0. We fail to reject H0.

### Q5 — Type I and Type II Errors

Even with a valid procedure, hypothesis tests can be wrong:

- **Type I Error (False Positive):** We reject H0, but H0 was actually true. We saw a pattern that was just random noise.
- **Type II Error (False Negative):** We fail to reject H0, but H0 was actually false. We missed a real effect.

The significance level directly controls the Type I error rate. At alpha = 0.05, we accept a 5% chance of a false positive.

## Probability

Probability measures how likely an event is, on a scale from 0 (impossible) to 1 (certain).

**Key definitions:**

- **Independent Events:** Two events where knowing one happened tells you nothing about the other. Rule: P(A and B) = P(A) * P(B).
- **Dependent Events:** The outcome of one event affects the probability of the other.
- **Mutually Exclusive Events:** Two events that cannot both happen at the same time. Rule: P(A or B) = P(A) + P(B).

**Probability rules:**

| Rule | Formula |
|---|---|
| Complement | P(not A) = 1 - P(A) |
| Addition (mutually exclusive) | P(A or B) = P(A) + P(B) |
| Addition (general) | P(A or B) = P(A) + P(B) - P(A and B) |
| Multiplication (independent) | P(A and B) = P(A) * P(B) |

**Conditional Probability:**

P(B | A) is read as "the probability of B given A already happened."

    P(B | A) = P(A and B) / P(A)

## Advanced Probability

### Bayes' Rule

Bayes' Rule lets us update our beliefs when we receive new information:

    P(A | B) = P(B | A) * P(A) / P(B)

Where:
- P(A) is the **prior** — how likely we thought A was before observing B
- P(B | A) is the **likelihood** — how likely B is if A is true
- P(A | B) is the **posterior** — how likely A is after observing B

The denominator P(B) is computed using the **law of total probability**:

    P(B) = P(B | A) * P(A) + P(B | not A) * P(not A)

Bayes' Rule is the foundation of many machine learning models and is widely used in statistics, medicine, and AI.
